```python
#непрерывные признаки
continuous_features = [
    'Age',
    'ФВ ЛЖ',
    'КДР ЛЖ',
    'КСР ЛЖ',
    'МЖП',
    'ЗСЛЖ',
    'ПСПЖ',
    'СДЛА',
    'Размер правого желудочка',
    'НПВ диаметр',
    'La1',
    'La2',
    'Ra1',
    'Ra2',
    'Гематокрит',
    'Гемоглобин',
    'Эритроциты',
    'Лейкоциты',
    'Тромбоциты',
    'Нейтрофилы (абсолютное значение)',
    'Нейтрофилы (относительное значение)',
    'Лимфоциты (абсолютное значение)',
    'Лимфоциты (относительное значение)',
    'Моноциты (абсолютное значение)',
    'Моноциты (относительное значение)',
    'Эозинофилы (абсолютное значение)',
    'Эозинофилы (относительное значение)',
    'Базофилы (абсолютное значение)',
    'Базофилы (относительное значение)',
    'Сумма всех лейкоцитов',
    'Средний объем эритроцита',
    'Среднее содержание гемоглобина в эритроците',
    'Ср, концентрация гемоглобина в эритроците',
    'Распределение эритроцитов по объему (RDW-CV)',
    'Распределение эритроцитов по объему (RDW-SD)',
    'Средний объем тромбоцита (MPV)',
    'Расчетная ширина распределения тромбоцитов',
    'Тромбокрит',
    'Коэффициент крупных тромбоцитов',
    'СОЭ',
    #'С-реактивный белок',
    'PLR (тромбоциты/лимфоциты абс) (61-239)',
    'SII (тромбоциты*нейтрофилы абс / лимфоциты абс)',
    'ELR (эозинофилы абс/лимфоциты абс',
    'ELR %',
    'ENR (эозин абс/нейтрофилы абс)',
    'BLR (базофилы абс/лимфоциты абс)',
    'GLR (глюкоза мг/дл / лимф абс)',
    'FLR (fibr/L abs)',
    'RLR (RDW/ lymph abs)',
    'MPV/PC (MPV/ PLT)',
    'Глюкоза в мг/дл',
    'Фибриноген',
    'Мочевина(b)',
    'Продолжительность операции',
    'Apache II',
    'Рост',
    'Вес',
    'Индекс массы тела',
    'Систолическое АД(b)',
    'Диастолического АД(b)',
    'ЧСС (b)',
    'Лейкоциты(a)',
    'Нейтрофилы (абсолютное значение)(a)',
    'Эр(a)',
    'Hb(a)',
    'Гематокрит(a)',
    'Тромбоциты(a)',
    'Глюкоза(a)',
    'Мочевина(a)',
    'ПТИ(a)',
    'МНО(a)',
    'ФБГ(a)',
    'ТВ(a)',
    'АПТВ(a)',
    'Креатинин',
    'СКФ',
    'SpO2',
    'GRACE(Рассчет)',
    'eGFR',
    #новые
    'АСАТ(a)',
    'АЛАТ(a)',
    'Билирубин общий(a)',
    'Холестерин общий(a)',
    'Триглицериды(a)',
    'Креатинин(a)',
    'ЛДГ(a)',
    'КФК(a)'
    #'MnSistAD',
    #'MnDiasAD',
    #'MnMeanAD',
    #'MxSistAD',
    #'MxDiasAD',
    #'MxMeanAD',
    #'SisStart',
    #'DiaStart',
    #'MeanStart',
    #'SisEnd',
    #'DiaEnd',
    #'MeanEnd'
]

#исправляем тип непрерывного признака Фибриноген
df_need= df_need.copy()
df_need.loc[:, 'Фибриноген'] = pd.to_numeric(df_need['Фибриноген'], errors='coerce')

#дихотомические признаки
dichotomous_features = [
    'Диагноз Нестабильная стенакордия',
    'ХСН_фк_10',
    'ХСН_фк_234',
    'Sex', # 1-мужчина 0 -женщина
    'Вид STEMI',#задний передний
    'Поражение ствола',#да нет
    'Инфаркт миокарда в анамнезе (<3)',#да нет
    'Инфаркт миокарда в анамнезе (>3)',#да нет
    'Инфаркт миокарда со стентированием в анамнезе',#да нет  и тут есть пустые
    'ОНМК (иш) в анамнезе',
    'ФП b (после чкв)', #да нет
    'ФП a (в анамнезе)',#да нет
    'Стенокардия (ИБ)',#да нет  и тут есть пустые
    'СД',#да нет
    'ХБП',#да нет и тут есть пустые
    'МКБ',
    'Стеноз аортального клапана (степень)',
    'Смерть',#да нет
    'повторные операции',
    'Смерть в реанимации',
    'GRACE(Общее)>140',
    'интегрилин|эптифибатид|коромакс|агграстат',#Да Нет и пустые
    'Общий анализ крови раньше операции',#Да Нет и пустые
    'Общий анализ крови_экспресс раньше операции',#Да Нет и пустые
    'Аортальный стеноз (Градиент)',
    'Аортальный стеноз (Заключение)',
    'ХОБЛ (Новый столбец)',#Да Нет и пустые
    'Стенокардия в диагнозе при поступлении',#Да Нет и пустые
    'ПИКС в диагнозе при поступлении',#Да Нет
    'Инфаркт-зависимая артерия(Огригированная)2',#Передний/Задний бассейн
    'Ишемический инсульт (Выгрузка из ИБ - временный столбец)',#Да Нет
    'Геморрагический инсульт (Выгрузка из ИБ - временный столбец)',#Да Нет
    'Инсульт (Выгрузка из ИБ - временный столбец)',#Да Нет
    'Курит',#Да Нет
    'GRACE(Общее)>156',
    'Отек легких(b)',
    'Отек легких(a)',#Да Нет
    'Отек легких',#Да Нет
    'Медицинская помощь оказана за первые 4 часа',#Да Нет
    'Гипертоническая болезнь',#Да Нет
    'GRACE (Низкий риск)',#Да Нет
    'GRACE (Средний риск)',#Да Нет
    'GRACE (Высокий риск)',#Да Нет
    'TIMI (Летальность) (Низкий риск)',#Да Нет
    'TIMI (Летальность) (Средний риск)',#Да Нет
    'TIMI (Летальность) (Высокий риск)',#Да Нет
    'PAMI (Низкий риск)',#Да Нет
    'PAMI (Средний риск)',#Да Нет
    'PAMI (Высокий риск)',#Да Нет
    'CADILLAC (Низкий риск)',
    'CADILLAC (Средний риск)',
    'CADILLAC (Высокий риск)',
    'РЕКОРД (Низкий риск)',#Да Нет
    'РЕКОРД (Высокий риск)',#Да Нет
    'А-В блокаДа (Новый)',#Да Нет
    'Адреналин',#Да Нет
    'Добутамин',#Да Нет
    'Допмин',#Да Нет
    'Норадреналин',#Да Нет
    'Левосимендан',#Да Нет
    'Мезатон',#Да Нет
    'Экстренный',#не меняем
    'Интегреллин',#не меняем
    'Тромболизис',
    #'Тромболизис у нас',#не меняем
    'ФВ<30%',#не меняем
    'Тирофибан',#не меняем
    'Аортальный стеноз (Градиент)',
    'Аортальный стеноз (Заключение)'
]
#'А-В блокада' - это можно не использовать так как есть признак 'А-В блокаДа (Новый)' который на 2 значения больше заполнен и уже является дихотомическим
'''def categorize_AB(value):
    if value == 'Нет':
        return 0
    else:
        return 1

df_need= df_need.copy()
df_need['А-В блокада категория'] = df_need['А-В блокада'].apply(categorize_AB)'''

#Исправляем столбец РЕКОРД и CADILLAC
#Создаем mapping для замены
_mapping = {
        'Нет': 0,
        'Да': 1,
        '0': 0,
        '1': 1,
        0: 0,
        1: 1
    }

# Заменяем значения
df_need['РЕКОРД (Низкий риск)'] = df_need['РЕКОРД (Низкий риск)'].replace(_mapping)
df_need['РЕКОРД (Высокий риск)'] = df_need['РЕКОРД (Высокий риск)'].replace(_mapping)

df_need['CADILLAC (Низкий риск)'] = df_need['CADILLAC (Низкий риск)'].replace(_mapping)
df_need['CADILLAC (Средний риск)'] = df_need['CADILLAC (Средний риск)'].replace(_mapping)
df_need['CADILLAC (Высокий риск)'] = df_need['CADILLAC (Высокий риск)'].replace(_mapping)


# Обогащаем парные признаками, заполняя пропуски в одном значением из другого
df_need['Курит'] = df_need['Курит'].fillna(df_need['Smoking'])
df_need['Тромболизис_x'] = df_need['Тромболизис_x'].fillna(df_need['Тромболизис_y'])
# Заменяем все числовые значения на соответствующие строки
df_need['Тромболизис_x'] = df_need['Тромболизис_x'].replace({
    1: 'Да',
    0: 'Нет',
    '1': 'Да',
    '0': 'Нет'
})
# Переименовываем объединенные признаки
df_need = df_need.rename(columns={'Тромболизис_x': 'Тромболизис'})

# Закодируем вручную дихотомические признаки с учетом особенностей
for col in dichotomous_features:
    if col in df_need.columns:
        unique_vals = df_need[col].dropna().unique()

        if set(unique_vals) == {'Да', 'Нет'}:
            # Особые случаи где "Нет" = 1, "Да" = 0
            if col in ['Тромболизис', 'Стенокардия (ИБ)', 'МКБ', 'Общий анализ крови_экспресс раньше операции',
                      'Инфаркт-зависимая артерия(Огригированная)2', 'Курит',
                      'Медицинская помощь оказана за первые 4 часа', 'Гипертоническая болезнь',
                      'GRACE (Низкий риск)', 'GRACE (Средний риск)',
                      'TIMI (Летальность) (Низкий риск)', 'TIMI (Летальность) (Средний риск)',
                      'PAMI (Низкий риск)', 'PAMI (Средний риск)',
                      'CADILLAC (Низкий риск)', 'CADILLAC (Средний риск)',
                      'РЕКОРД (Низкий риск)', 'Экстренный']:
                df_need[col] = df_need[col].replace({'Нет': 1, 'Да': 0})
            else:
                df_need[col] = df_need[col].replace({'Да': 1, 'Нет': 0})

        elif set(unique_vals) == {'М', 'Ж'}: #Пол
            df_need[col] = df_need[col].replace({'М': 0, 'Ж': 1})

        elif set(unique_vals) == {'Задний', 'Передний'}: #Вид STEMI
            df_need[col] = df_need[col].replace({'Задний': 0, 'Передний': 1})

        elif set(unique_vals) == {'Правый бассейн', 'Передний бассейн'}:
            df_need[col] = df_need[col].replace({'Правый бассейн': 1, 'Передний бассейн': 0}) #Инфаркт-зависимая артерия(Огригированная)2


#категориальные признаки
categorical_features=[
    'Инфаркт-зависимая артерия',
    'Количество пораженных сосудов(Syntax)',
    'Количество пораженных сосудов(Значимость)',
    'TIMI категория',
    'CADILLAC категория',
    'РЕКОРД категория',
    'PAMI категория',
    'TIMI Летальность категория',
    #'ХСН фк',
    'Вид операции(ИБ)(Новый)',
    'Исход заболевания',
    #'Сопутствующие1',
    'Инфаркт-зависимая артерия(Огригированная)',
    'KDIGO',
    'Baun',
    'MKB категория',
    'APACHE2 выписка категория',
    'APACHE2 1сутки категория',
    'APACHE2 3дня категория',
    'Экстренная операция(ИБ)(Новый)', #экстренная/плановая и Да
    'ГБ стадия категория',
    'ГБ риск категория'

]


#переделаем некоторые категориальные признаки
#МКВ
def categorize_MKB(value):
    if value in ['C18.9', 'C25.2',  'C34.3', 'C34.9', 'C50.0', 'C67.9', 'C90.0', 'D33.3']:
        return 'КЛАСС III БОЛЕЗНИ КРОВИ, КРОВЕТВОРНЫХ ОРГАНОВ И ОТДЕЛЬНЫЕ НАРУШЕНИЯ, ВОВЛЕКАЮЩИЕ ИММУННЫЙ МЕХАНИЗМ (D50-D89)'
    elif value in ['E05.0', 'E10.0', 'E10.6', 'E10.7', 'E11.6', 'E11.7','E13.7'  ]:
        return 'КЛАСС IV БОЛЕЗНИ ЭНДОКРИННОЙ СИСТЕМЫ, РАССТРОЙСТВА ПИТАНИЯ И НАРУШЕНИЯ ОБМЕНА ВЕЩЕСТВ (E00-E90)'
    elif value in ['G09.0', 'G23.8', 'G25.5', 'G31.0', 'G37.8', 'G45.0', 'G46.0',  'G55.1', 'G55.3', 'G56.1', 'G90.0', 'G93.4' ]:
        return 'КЛАСС VI БОЛЕЗНИ НЕРВНОЙ СИСТЕМЫ (G00-G99)'
    elif value in ['I20.0','I20.00','I20.8','I20.80','I20.9', 'I21.0', 'I21.00', 'I21.1', 'I21.10', 'I21.2', 'I21.3',  'I21.4',  'I21.40',
                 'I21.9', 'I21.90', 'I22.0', 'I22.00', 'I22.1', 'I22.10','I22.8', 'I22.80', 'I22.9', 'I22.90', 'I23.8','I24.0', 'I24.8', 'I25.0', 'I25.1', 'I25.2', 'I25.5',  'I25.50', 'I25.6', 'I25.8', 'I26.0', 'I26.9', 'I27.0', 'I27.9', 'I31.0', 'I32.0', 'I33.0', 'I33.9', 'I34.0', 'I34.2', 'I35.0' , 'I35.1', 'I35.2', 'I39.0','I40.0', 'I41.8', 'I42.0', 'I42.1', 'I42.2', 'I42.7', 'I42.8', 'I42.9', '143.0', '143.1', '144.0', '144.2', '144.3', '144.7', '145.2', '145.5' , '145.6', '145.8', '147.0' , '147.1', '147.2', '147.9', '148.0', '148.1', '148.2', '148.9', '149.0', '149.3', '149.4', '149.5', '149.8', '149.9', '150.0', '150.9', '151.9', 'I61.0', 'I61.1', 'I61.3', 'I63.3', 'I63.4', 'I61.40', 'I63.40', 'I63.9','I69.3', 'I63.5', 'I67.8', 'I67.9', 'I69.0', 'I69.8', 'I69.80', 'I70.0', 'I70.2', 'I70.9', 'I73.8',  'I82.8', 'I83.9', 'I11.0', 'I11.9']:
        return 'КЛАСС IX БОЛЕЗНИ СИСТЕМЫ КРОВООБРАЩЕНИЯ (I00-I99)'
    elif value in ['J11.1','J84.0', 'J47.0', 'J15.4', 'J18.0', 'J90.0', 'J45.0', 'J44.0', 'J44.9', 'J84.9', 'J45.9', 'J13.0' ]:
        return 'КЛАСС X БОЛЕЗНИ ОРГАНОВ ДЫХАНИЯ (J00-J99)'
    elif value in [ 'K80.0', 'K86.1', 'K80.1','K50.1', 'K57.2', 'K63.2', 'K82.8','K52.9', 'K73.8', 'K25.4', 'K74.0', 'K22.1','K57.3','K74.6','K26.2', 'K51.1','K85.0','K21.0', 'K58.9', 'K26.0', 'K56.0', 'K51.0', 'K82.9' ]:
        return 'КЛАСС XI БОЛЕЗНИ ОРГАНОВ ПИЩЕВАРЕНИЯ (K00-K93)'
    elif value in [ 'M07.0', 'M16.1', 'M15.0', 'M42.1', 'M05.0', 'M06.0',  'M02.9', 'M05.3', 'M33.2', 'M02.0', 'M42.0', 'M48.9', 'M34.9', 'M51.0', 'M45.0']:
        return 'КЛАСС XIII БОЛЕЗНИ КОСТНО-МЫШЕЧНОЙ СИСТЕМЫ И СОЕДИНИТЕЛЬНОЙ ТКАНИ (M00-M99)'
    elif value in [ 'N20.1', 'N83.9', 'N73.9', 'N20.9', 'N40.0', 'N11.0', 'N13.3', 'N83.1',  'N31.2']:
        return 'КЛАСС XIV БОЛЕЗНИ МОЧЕПОЛОВОЙ СИСТЕМЫ (N00-N99)'
    elif value in ['S06.3', 'S06.6', 'S30.0', 'S06.0', 'S82.6', 'S72.0',  'S82.8', 'T84.1' ]:
        return 'КЛАСС XIX ТРАВМЫ, ОТРАВЛЕНИЯ И НЕКОТОРЫЕ ДРУГИЕ ПОСЛЕДСТВИЯ ВОЗДЕЙСТВИЯ ВНЕШНИХ ПРИЧИН (S00-T98)'
    else:
        'Другие заболевания'

df_need['MKB категория'] = df_need['MKB'].apply(categorize_MKB)
#точно правильно




def categorize_GV_st(value):
    if value in [0.0, 1.0]:
        return 'Низкий уровень'
    elif value in [2.0, 3.0]:
        return 'Средний уровень'
    elif value >= 4.0:
        return 'Высокий уровень'
    else:
        return 'Неизвестно'

df_need['ГБ стадия категория'] = df_need['ГБ стадия'].apply(categorize_GV_st)


def categorize_GV_r(value):
    if value in [0.0, 1.0, 2.0]:
        return 'Низкий уровень'
    elif value in [3.0, 4.0, 5.0]:
        return 'Средний уровень'
    elif value >= 6.0:
        return 'Высокий уровень'
    else:
        return 'Неизвестно'

df_need['ГБ риск категория'] = df_need['ГБ риск'].apply(categorize_GV_r)

def categorize_cadillac(value):
    if value in [0.0, 1.0, 2.0]:
        return 'Низкий уровень'
    elif value in [3.0, 4.0, 5.0]:
        return 'Средний уровень'
    elif value >= 6.0:
        return 'Высокий уровень'
    else:
        return 'Неизвестно'

df_need['CADILLAC категория'] = df_need['CADILLAC'].apply(categorize_cadillac)


def categorize_record(value):
    if value in [0.0, 1.0]:
        return 'Низкий риск'
    elif value in [2.0]:
        return 'Умеренный риск'
    elif value in [3.0]:
        return 'Высокий риск'
    elif value >= 4.0:
        return 'Очень высокий'
    else:
        return 'Неизвестно'

df_need['РЕКОРД категория'] = df_need['РЕКОРД'].apply(categorize_record)



def categorize_pami(value):
    if value in [0, 1, 2, 3]:
        return 'Низкий риск'
    elif value in [4, 5, 6]:
        return 'Средний риск'
    elif value >= 7:
        return 'Высокий риск'
    else:
        return 'Неизвестно'

df_need['PAMI категория'] = df_need['PAMI'].apply(categorize_pami)


def timi_to_binary(value):
    """
    Преобразует TIMI в бинарный признак для модели.
    Возвращает:
    0 — неуспешная реперфузия (высокий риск)
    1 — успешная реперфузия (низкий риск)
    """
    if value in [0.0, 1.0, 2.0]:
        return 0  # Неуспех / высокий риск
    elif value == 3.0:
        return 1  # Успех / низкий риск
    else:
        return None  # пропуск или ошибка


def categorize_timi_deth(value):
    if value in [0.0, 1.0, 2.0]:
        return 'Низкий риск'
    elif value in [3.0, 4.0]:
        return 'Умеренный риск'
    elif value in [5.0, 6.0]:
        return 'Высокий риск'
    elif value >= 7.0 :
        return 'Очень высокий риск'
    else:
        return 'Неизвестно'

df_need['TIMI категория'] = df_need['TIMI'].apply(timi_to_binary)
df_need['TIMI Летальность категория'] = df_need['TIMI (Летальность)'].apply(categorize_timi_deth)



def categorize_apache2(score):
    if score <= 10:
        return 'Низкий риск (0-10)'
    elif 11 <= score <= 16:
        return 'Умеренный риск (11-16)'
    elif 17 <= score <= 24:
        return 'Высокий риск (17-24)'
    elif score > 24:
        return 'Критический риск (25+)'
    else:
        return 'Неизвестно'
#ЕА проверил
df_need= df_need.copy()
df_need['APACHE2 выписка категория'] = df_need['APACHE2 выписка'].apply(categorize_apache2)
df_need['APACHE2 1сутки категория'] = df_need['APACHE2 1сутки'].apply(categorize_apache2)
df_need['APACHE2 3дня категория'] = df_need['APACHE2 3дня'].apply(categorize_apache2)




df_need = df_need.rename(columns={'Age':'Возраст',
                        'Sex':'Пол',
                        'Инсульт (Выгрузка из ИБ - временный столбец)': 'Инсульт',
                        'Ишемический инсульт (Выгрузка из ИБ - временный столбец)':'Ишемический инсульт',
                        'Экстренная операция(ИБ)(Новый)':'Экстренная операция',
                        'ХОБЛ (Новый столбец)':'ХОБЛ',
                        'Стенокардия (ИБ)':'Стенокардия',
                        'Геморрагический инсульт (Выгрузка из ИБ - временный столбец)':'Геморрагический инсульт',
                        'А-В блокаДа (Новый)':'А-В блокаДа'})
```
