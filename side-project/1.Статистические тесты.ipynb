```python
#Таргет
target = 'КШ развился в реанимации'

#Работаем с непрерывными признаками и применяем к ним тест манна уитни

#словарь для хранения результатов
results_con = {}

#проверяем, является ли столбец непрерывным признаком
for column in continuous_features:
    group1 = df[df[target] == 1][column]
    group0 = df[df[target] == 0][column]

    group1 = pd.to_numeric(group1, errors='coerce')
    group0 = pd.to_numeric(group0, errors='coerce')

    group1 = group1.dropna()
    group0 = group0.dropna()

    if not group0.empty and not group1.empty:
        print(f"Статистика для {column} в group0:")
        print(group0.count())
        print(f"Статистика для {column} в group1:")
        print(group1.count())

        #тест Манна-Уитни
        stat_mw, p_value_mw = stats.mannwhitneyu(group1, group0)


        #медианы для каждой группы
        median_group1 = round(group1.median(), 2)
        median_group0 = round(group0.median(), 2)

        #Q1 Q3 для каждой группы
        quartiles_group0 = np.percentile(group0, [25, 75])  # Q1 и Q3
        quartiles_group1 = np.percentile(group1, [25, 75])  # Q1 и Q3
        quartiles_group0 = np.round(quartiles_group0, 3)
        quartiles_group1 = np.round(quartiles_group1, 3)

        #проверяем на NaN перед добавлением в результаты
        if not (pd.isna(stat_mw) or pd.isna(p_value_mw)):
            results_con[column] = {
                'Null' : df[column].isnull().sum(),
                'Not Null' : df.shape[0] -df[column].isnull().sum(),
                'Median' : df[column].median(),
                'Median_1(Q1;Q3)': f'{median_group1}({quartiles_group1[0]};{quartiles_group1[1]})',
                'Median_0(Q1;Q3)': f'{median_group0}({quartiles_group0[0]};{quartiles_group0[1]})',
                'Mann-Whitney p-value': p_value_mw
                #,'Mann-Whitney Statistic': stat_mw
            }
    else:
        print(f"Нет данных для {column} в одной из групп.")

#создание DataFrame из результатов
results_df = pd.DataFrame.from_dict(results_con, orient='index')

#сохранение DataFrame в Excel файл
results_df.to_excel('непрерывные_признаки_09_03_26.xlsx', sheet_name='Результаты', index=True)

print("Результаты успешно сохранены в файл непрерывные_признаки 09_03_26.xlsx")










#для дихотомических признаков тест фишера
results_dih = {}

#проверяем, является ли столбец дихотомическим признаком (0 и 1 и др)
for column in dichotomous_features:
    temp_df = df[[target, column]].dropna() #удаляем пустые значения
    #создаем таблицу сопряженности
    ct = pd.crosstab(temp_df.loc[:, (target)], temp_df.loc[:, (column)])
    ct.index = ['Не развился', 'Развился']
    ct.columns = ['0', '1']
    total_rows = len(df)  # всего строк в исходных данных
    null_count = df[column].isnull().sum()  # пропуски именно в этом признаке
    not_null_count = total_rows - null_count  # непустых значений

    #проверяем, что таблица 2x2
    if ct.shape == (2, 2) and (ct.values > 0).all():
        #вычисляем отношение шансов и p-value (тест Фишера)
        _, pvalue = stats.fisher_exact(ct)

        table = sm.stats.Table2x2(np.asarray(ct))
        #извлекаем нужные показатели:
        OR = table.oddsratio  #OR
        ci = table.oddsratio_confint()  #доверительный интервал 95%
        (lower_ci, upper_ci) = ci  #границы интервала

        #РАСЧЕТ АБСОЛЮТНЫХ И ОТНОСИТЕЛЬНЫХ РАСПРЕДЕЛЕНИЙ
        #Абсолютные значения
        cat0_no_event = ct.loc['Не развился', '0']  # категория 0 без события
        cat0_with_event = ct.loc['Развился', '0']   # категория 0 с событием
        cat1_no_event = ct.loc['Не развился', '1']  # категория 1 без события
        cat1_with_event = ct.loc['Развился', '1']   # категория 1 с событием

        #Общие суммы по группам (КШ и без КШ)
        total_no_event = cat0_no_event + cat1_no_event  # всего без КШ
        total_with_event = cat0_with_event + cat1_with_event  # всего с КШ

        #Относительные значения для ПРИЗНАКА = 1 (внутри каждой группы)
        # Для группы "Без КШ" - процент пациентов с признаком = 1
        if total_no_event > 0:
            rel_cat1_no_event = cat1_no_event / total_no_event  # доля признака=1 среди без КШ
        else:
            rel_cat1_no_event = 0

        # Для группы "С КШ" - процент пациентов с признаком = 1
        if total_with_event > 0:
            rel_cat1_with_event = cat1_with_event / total_with_event  # доля признака=1 среди с КШ
        else:
            rel_cat1_with_event = 0

        #Абсолютные значения для ПРИЗНАКА = 1
        abs_cat1_no_event = cat1_no_event  # абсолютное число признака=1 среди без КШ
        abs_cat1_with_event = cat1_with_event  # абсолютное число признака=1 среди с КШ

        #ХИ-КВАДРАТ
        temp_df = df[[target, column]].dropna()

        # Создаем таблицу сопряженности
        ct_chi = pd.crosstab(temp_df[target], temp_df[column])

        # Вычисляем хи-квадрат
        chi2, p_value, dof, expected = stats.chi2_contingency(ct_chi)



        #сохраняем результаты
        results_dih[column] = {
            'Null' : null_count,
            'Not Null' : not_null_count,
            'F-тест p-value': pvalue,
            'Chi2-тест p-value': p_value,
            'OR [95% CI]': f'{OR:.2f} [{lower_ci:.2f};{upper_ci:.2f}]',
            # Абсолютные значения для признака = 1
            'Без КШ (абс)': abs_cat1_no_event,
            'КШ (абс)': abs_cat1_with_event,
            # Относительные значения для признака = 1 (проценты)
            'Без КШ (%)': f'{rel_cat1_no_event:.3f}',
            'КШ (%)': f'{rel_cat1_with_event:.3f}',
            # Форматированные строки для удобства чтения
            'Распределение признака=1': f'Без КШ: {abs_cat1_no_event}/{total_no_event} ({rel_cat1_no_event:.1%}) | КШ: {abs_cat1_with_event}/{total_with_event} ({rel_cat1_with_event:.1%})'
        }

#создание DataFrame из результатов
results_df = pd.DataFrame.from_dict(results_dih, orient='index')

#сохранение DataFrame в Excel файл
results_df.to_excel('дихотомические_признаки_19_04_26.xlsx', sheet_name='Результаты', index=True)



#для категориальных признаков тест хи-квадрат
results_cat = {}

#проверяем, является ли столбец категориальным
for column in categorical_features:
    temp_df = df[[target, column]].dropna()

    # Создаем таблицу сопряженности
    ct = pd.crosstab(temp_df[target], temp_df[column])
    ct.index = ['Не развился', 'Развился']

    # Вычисляем хи-квадрат
    chi2, p_value, dof, expected = stats.chi2_contingency(ct)

    # Общая информация по признаку
    null_count =  df[column].isnull().sum()
    not_null_count = df.shape[0] - null_count

    # Сохраняем общую информацию
    results_cat[column] = {
        'Тип_записи': 'Общий_признак',
        'Null': null_count,
        'Not Null': not_null_count,
        'p_value_chi2': p_value,
        'Распределение': f"Всего категорий: {len(ct.columns)}",
        'Абс_без_события': '',
        'Абс_с_событием': '',
        'Всего_в_категории': '',
        'Отн_без_события': '',
        'Отн_с_событием': ''
    }

    # Добавляем информацию по каждой категории
    for category in ct.columns:
        # Абсолютные значения
        no_event_count = ct.loc['Не развился', category]
        event_count = ct.loc['Развился', category]
        total_category = no_event_count + event_count

        # Относительные значения
        if total_category > 0:
            no_event_percent = no_event_count / total_category
            event_percent = event_count / total_category
        else:
            no_event_percent = 0
            event_percent = 0

        # Сохраняем информацию по категории
        results_cat[f"{column}_{category}"] = {
            'Тип_записи': 'Категория',
            'Null': '',
            'Not Null': '',
            'p_value_chi2': '',
            'Распределение': f"{category}",
            'Абс_без_события': no_event_count,
            'Абс_с_событием': event_count,
            'Всего_в_категории': total_category,
            'Отн_без_события': f'{no_event_percent:.1%}',
            'Отн_с_событием': f'{event_percent:.1%}'
        }

#создание DataFrame из результатов
results_df = pd.DataFrame.from_dict(results_cat, orient='index')

#сохранение DataFrame в Excel файл
results_df.to_excel('категориальные_признаки_19_04_26.xlsx', sheet_name='Результаты', index=True)
```
