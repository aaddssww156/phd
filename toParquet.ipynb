# Обработка датафреймов

### Проблема

При работе с датафреймами очень много времени уходит на изначальную загрузку датасета из xlsx файла. Этот этап можно упростить, преобразовав датафрейм в другой тип хранения данных

```python
%pip install pyarrow
```

```python
import pandas as pd
```

```python
table_data = pd.read_excel("./data_raw/DataSet_V49 (2).xlsx")
```

```python
print(f"Табличные данные успешно загружены! Размер: {table_data.shape}")
print(f"Занимаемая память: {table_data.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
```

```python
table_data = table_data.convert_dtypes(dtype_backend='numpy_nullable')
```

```python
# Преобразуем все столбцы типа object в строки
for col in table_data.select_dtypes(include=['object']).columns:
    table_data[col] = table_data[col].astype(str)

# Сохраняем
table_data.to_parquet("./data_phd/table.parquet")
```

```python
table_data = None 

table_data = pd.read_parquet("./data_phd/table.parquet")
```

```python
print(f"Табличные данные успешно загружены! Размер: {table_data.shape}")
print(f"Занимаемая память: {table_data.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
```

```python
table_data.to_feather("./data_phd/table.feather")
```

```python
table_data = None 

table_data = pd.read_feather("./data_phd/table.feather")
```

```python
print(f"Табличные данные успешно загружены! Размер: {table_data.shape}")
print(f"Занимаемая память: {table_data.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
```
